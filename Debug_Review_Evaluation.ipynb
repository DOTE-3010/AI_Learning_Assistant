## Part 3: Evaluation (The AI Judge)

Testing code is easy; testing "intelligence" is hard. How do we know if Solver42's answer is **correct**?

We use an **Evaluation Agent**.

### 3.1 The "Web Search Verifier" Approach
We can ask an AI agent to:
1.  Read our app's output.
2.  Search the web for the ground truth.
3.  Compare and score the accuracy.

```python
# Example Output from Solver42 (Hypothetical)
generated_answer = "The capital of Australia is Sydney."

print(f"Checking verification for: '{generated_answer}'")

# INSTRUCTIONS FOR STUDENTS:
# 1. Copy the prompt below.
# 2. Open Cursor Chat (Agent Mode).
# 3. Paste the prompt to verify the fact above.

verification_prompt = f"""
I am evaluating an AI model's accuracy.
Statement: '{generated_answer}'

Task:
1. Use web search to find the actual capital of Australia.
2. Compare it with the Statement.
3. If incorrect, provide the correct answer.
4. Rate the statement's accuracy on a scale of 0 to 1.
"""

print("\n--- Copy this into Cursor Chat ---\n")
print(verification_prompt)
```

### 3.2 Automated Evaluation Script (LLM-as-a-Judge)
For batch processing, we can write a script to evaluate multiple answers.

```python
import os

def evaluate_answer(question, answer, ground_truth=None):
    """
    A skeleton function for LLM-as-a-Judge.
    In a real scenario, this would call the OpenAI API with a 'Judge Prompt'.
    """
    prompt = f"""
    Question: {question}
    Student Answer: {answer}
    Ground Truth: {ground_truth}
    
    Rate the Student Answer from 1-5 based on correctness.
    """
    print(f"[System] Sending to Judge Model:\n{prompt}")
    # client.chat.completions.create(...) -> Score
    return "(Simulated Score: 3/5 - Correct entity but wrong location)"

# Try it
print(evaluate_answer(
    question="Capital of Australia", 
    answer="Sydney", 
    ground_truth="Canberra"
))
```
